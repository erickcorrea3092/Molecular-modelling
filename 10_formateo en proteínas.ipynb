# Visualización del hetfleg

```python
import py3Dmol

# --- Ejemplo 1: Mostrando los heteroátomos (ligando y agua) ---
# hetflag=True es el comportamiento por defecto, pero lo ponemos para ser explícitos.
view1 = py3Dmol.view(query='pdb:1a28', options={'hetflag': True})

# Estilos para la visualización
view1.setStyle({'cartoon': {'color': 'blue'}}) # Colorea la proteína como un arcoíris #azul
view1.setStyle({'hetflag': True}, {'stick': {'colorscheme': 'yellowCarbon', 'radius': 0.8}}) # Muestra el ligando en estilo de varillas #amarillo
view1.setStyle({'hetflag': True, 'resn': 'HOH'}, {'sphere': {'radius': 0.5, 'color': 'red'}}) # Muestra las moléculas de agua como esferas rojas #rojo
# Centramos la vista en la molécula
view1.zoomTo()
```

# visualización sin agua

```python
import py3Dmol

# Crear una vista en 3D
view = py3Dmol.view(query='pdb:1a28', options={'hetflag': True},width=1590, height=900)

# Cargar la estructura de la proteína desde un archivo PDB
# Reemplaza '1A3N.pdb' con el nombre de tu archivo PDB


# Seleccionar todos los átomos que no son moléculas de agua y aplicar un estilo
view.setStyle({'not': {'resn': ['HOH']}}, {'cartoon': {'color': 'blue'}})
view.setStyle({'hetflag': True}, {'stick': {'colorscheme': 'yellowCarbon', 'radius': 0.8}}) # Muestra el ligando en estilo de varillas
# Ajustar la vista para centrar la proteína
view.zoomTo()

# Mostrar la visualización
view.show()
```

# Añadir etiquetas

```python
# Añadimos una etiqueta
view1.addLabel('PDB ID: 1A28',
    #"PDB ID: 1A28",
               {'position': {'x': -4, 'y': -4, 'z': 0},
                'fontColor': 'black',
                'backgroundColor': 'lightgray',
                'backgroundOpacity': 0.8})

# Mostramos la visualización
print("Visualización 1: Proteína con su ligando y moléculas de agua.")
view1.show()
```

```python
# --- Ejemplo 2: Ocultando los heteroátomos ---
# Aquí establecemos hetflag=False para ocultar ligandos y aguas al inicio.
view2 = py3Dmol.view(query='pdb:1A28', options={'hetflag': False})

# Aplicamos estilo solo a la proteína
view2.setStyle({'cartoon': {'color': 'spectrum'}})

# Centramos la vista
view2.zoomTo()

# Añadimos una etiqueta
view2.addLabel("Con hetflag=False: Se ocultan ligando y aguas",
               {'position': {'x': -4, 'y': -4, 'z': 0},
                'fontColor': 'black', 'backgroundColor':
                'lightgray', 'backgroundOpacity': 0.8})

# Mostramos la visualización
print("Etiqueta:\nVisualización 2: Solo la estructura de la proteína.")
view2.show()
```

```python
import py3Dmol

# Crear el visor
viewer = py3Dmol.view(query='pdb:1a28', width=1590, height=900) # 1UBQ, ubiquitina, 1BNA ADN, 4HHB hemoglobin,

# Estilo cartoon para la estructura secundaria
#viewer.setStyle({'cartoon': {'color': 'spectrum'}})

viewer.setStyle({'cartoon': {'color': 'yellow'}})

# Estilo para el ligando
viewer.addStyle({'hetflag': True},
                {'stick': {'colorscheme': 'redCarbon'}})


# Añadir superficie transparente
viewer.addSurface(py3Dmol.VDW, {'opacity': 0.3, 'color': 'grey'})


# Resaltar residuos hidrofóbicos (ejemplo: leucina, valina)
viewer.addStyle({'resi': [724, 776, 788, 759, 760], 'chain': 'A'},
                {'stick': {'colorscheme': 'greenCarbon'}})

# Añadir etiquetas
viewer.addResLabels({'resi': [724, 776, 778, 759, 760], 'chain': 'A'},
                    {'fontColor': 'white', 'backgroundColor': 'black'})

# Centrar y mostrar
viewer.zoomTo()
viewer.setBackgroundColor('white')
viewer.show()
```
